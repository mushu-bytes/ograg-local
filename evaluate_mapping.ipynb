```python
import pickle
import altair
import pandas as pd
import numpy as np
```

```python
low_level_ont_mapping = pickle.load(open('ontology_mapping_evaluation/low_level_mapping_11_8.pkl', 'rb'))
high_level_ont_mapping = pickle.load(open('ontology_mapping_evaluation/high_level_mapping_11_8.pkl', 'rb'))
full_ont_mapping = pickle.load(open('ontology_mapping_evaluation/full_mapping_11_8.pkl', 'rb'))

low_level_ont_mapping_no_defs = pickle.load(open('ontology_mapping_evaluation/low_level_mapping_no_defs.pkl', 'rb'))
high_level_ont_mapping_no_defs = pickle.load(open('ontology_mapping_evaluation/high_level_mapping_no_defs.pkl', 'rb'))
full_ont_mapping_no_defs = pickle.load(open('ontology_mapping_evaluation/full_mapping_no_defs.pkl', 'rb'))

sci_subcategories_ont_mapping = pickle.load(open('ontology_mapping_evaluation/sci_subcategories_mapping.pkl', 'rb'))
sci_categories_ont_mapping = pickle.load(open('ontology_mapping_evaluation/sci_categories_mapping.pkl', 'rb'))
sci_full_ont_mapping = pickle.load(open('ontology_mapping_evaluation/sci_full_mapping.pkl', 'rb'))
```

```python
# Prepare data in long format for Altair
data = [
    {
        'Domain': 'CLBP',
        'Level': 'Fine-grained',
        'Score': np.mean(low_level_ont_mapping_no_defs['results']['nv_response_groundedness'])
    },
    {
        'Domain': 'CLBP',
        'Level': 'Coarse-grained',
        'Score': np.mean(high_level_ont_mapping_no_defs['results']['nv_response_groundedness'])
    },
    {
        'Domain': 'CLBP',
        'Level': 'Full Ontology',
        'Score': np.mean(full_ont_mapping_no_defs['results']['nv_response_groundedness'])
    },
    {
        'Domain': 'SCI',
        'Level': 'Fine-grained',
        'Score': np.mean(sci_subcategories_ont_mapping['results']['nv_response_groundedness'])
    },
    {
        'Domain': 'SCI',
        'Level': 'Coarse-grained',
        'Score': np.mean(sci_categories_ont_mapping['results']['nv_response_groundedness'])
    },
    {
        'Domain': 'SCI',
        'Level': 'Full Ontology',
        'Score': np.mean(sci_full_ont_mapping['results']['nv_response_groundedness'])
    }
]

df = pd.DataFrame(data)
df
```

```python
# Create grouped bar chart with labels
base = altair.Chart(df).encode(
    x=altair.X('Level:N', title='Ontology Level', axis=altair.Axis(labelFontWeight='bold')),
    y=altair.Y('Score:Q', title='Response Groundedness', scale=altair.Scale(domain=[0, 1.2]), axis=altair.Axis(grid=False)),
    color=altair.Color('Level:N', title=None, legend=None, scale=altair.Scale(scheme='dark2'))
)

bars = base.mark_bar()

text = base.mark_text(
    align='center',
    baseline='bottom',
    dy=-5,
    fontWeight='bold',
    color='black',
    fontSize=14
).encode(
    text=altair.Text('Score:Q', format='.3f')
)

chart = (bars + text).properties(
    width=200,
    height=250
).facet(
    column=altair.Column('Domain:N', title='Domain', header=altair.Header(labelFontWeight='bold'))
)

chart
```

```python
chart.save('evaluation/figs/ont_mapping_domains_comparison.png', ppi=500)
```
