```python
%load_ext autoreload
%autoreload 2

import pickle
import pandas as pd
import numpy as np
import altair
```

```python
low_level_ontology_eval = pickle.load(open('ragas_eval/low_level_ontology_eval_11_8.pkl', 'rb'))
high_level_ontology_eval = pickle.load(open('ragas_eval/high_level_ontology_eval_11_8.pkl', 'rb'))
full_ontology_eval = pickle.load(open('ragas_eval/full_ontology_eval_11_8.pkl', 'rb'))
rag_eval = pickle.load(open('ragas_eval/rag_eval_11_8.pkl', 'rb'))

low_level_ontology_eval_no_defs = pickle.load(open('ragas_eval/low_level_ontology_eval_no_defs.pkl', 'rb'))
high_level_ontology_eval_no_defs = pickle.load(open('ragas_eval/high_level_ontology_eval_no_defs.pkl', 'rb'))
full_ontology_eval_no_defs = pickle.load(open('ragas_eval/full_ontology_eval_no_defs.pkl', 'rb'))

sci_categories_eval = pickle.load(open('ragas_eval/sci_categories_eval.pkl', 'rb'))
sci_subcategories_eval = pickle.load(open('ragas_eval/sci_subcategories_eval.pkl', 'rb'))
sci_full_eval = pickle.load(open('ragas_eval/sci_full_eval.pkl', 'rb'))
sci_rag_eval = pickle.load(open('ragas_eval/sci_rag_eval.pkl', 'rb'))
```

```python
# Prepare data for Response Groundedness chart
groundedness_data = [
    {
        'Domain': 'CLBP',
        'Ontology': 'Full Ontology',
        'Score': np.mean(full_ontology_eval_no_defs['results']['nv_response_groundedness'])
    },
    {
        'Domain': 'CLBP',
        'Ontology': 'Fine-grained',
        'Score': np.mean(low_level_ontology_eval_no_defs['results']['nv_response_groundedness'])
    },
    {
        'Domain': 'CLBP',
        'Ontology': 'Coarse-grained',
        'Score': np.mean(high_level_ontology_eval_no_defs['results']['nv_response_groundedness'])
    },
    {
        'Domain': 'CLBP',
        'Ontology': 'RAG (No Ontology)',
        'Score': np.mean(rag_eval['results']['nv_response_groundedness'])
    },
    {
        'Domain': 'SCI',
        'Ontology': 'Full Ontology',
        'Score': np.mean(sci_full_eval['results']['nv_response_groundedness'])
    },
    {
        'Domain': 'SCI',
        'Ontology': 'Fine-grained',
        'Score': np.mean(sci_subcategories_eval['results']['nv_response_groundedness'])
    },
    {
        'Domain': 'SCI',
        'Ontology': 'Coarse-grained',
        'Score': np.mean(sci_categories_eval['results']['nv_response_groundedness'])
    },
    {
        'Domain': 'SCI',
        'Ontology': 'RAG (No Ontology)',
        'Score': np.mean(sci_rag_eval['results']['nv_response_groundedness'])
    }
]

groundedness_df = pd.DataFrame(groundedness_data)
groundedness_df
```

```python
# Prepare data for Context Relevance chart
relevance_data = [
    {
        'Domain': 'CLBP',
        'Ontology': 'Full Ontology',
        'Score': np.mean(full_ontology_eval_no_defs['results']['nv_context_relevance'])
    },
    {
        'Domain': 'CLBP',
        'Ontology': 'Fine-grained',
        'Score': np.mean(low_level_ontology_eval_no_defs['results']['nv_context_relevance'])
    },
    {
        'Domain': 'CLBP',
        'Ontology': 'Coarse-grained',
        'Score': np.mean(high_level_ontology_eval_no_defs['results']['nv_context_relevance'])
    },
    {
        'Domain': 'CLBP',
        'Ontology': 'RAG (No Ontology)',
        'Score': np.mean(rag_eval['results']['nv_context_relevance'])
    },
    {
        'Domain': 'SCI',
        'Ontology': 'Full Ontology',
        'Score': np.mean(sci_full_eval['results']['nv_context_relevance'])
    },
    {
        'Domain': 'SCI',
        'Ontology': 'Fine-grained',
        'Score': np.mean(sci_subcategories_eval['results']['nv_context_relevance'])
    },
    {
        'Domain': 'SCI',
        'Ontology': 'Coarse-grained',
        'Score': np.mean(sci_categories_eval['results']['nv_context_relevance'])
    },
    {
        'Domain': 'SCI',
        'Ontology': 'RAG (No Ontology)',
        'Score': np.mean(sci_rag_eval['results']['nv_context_relevance'])
    }
]

relevance_df = pd.DataFrame(relevance_data)
relevance_df
```

```python
combined_data = []

# Add groundedness data
for item in groundedness_data:
    combined_data.append({
        'Domain': item['Domain'],
        'Ontology': item['Ontology'],
        'Metric': 'Response Groundedness',
        'Score': item['Score']
    })

# Add relevance data
for item in relevance_data:
    combined_data.append({
        'Domain': item['Domain'],
        'Ontology': item['Ontology'],
        'Metric': 'Context Relevance',
        'Score': item['Score']
    })

combined_df = pd.DataFrame(combined_data)
combined_df
```

```python
base = altair.Chart(combined_df).encode(
    x=altair.X('Ontology:N', title='Ontology Level', axis=altair.Axis(labelFontWeight='bold')),
    y=altair.Y('Score:Q', title=None, scale=altair.Scale(domain=[0, 1]), axis=altair.Axis(grid=False)),
    color=altair.Color('Ontology:N', title='Ontology Level', legend=None, scale=altair.Scale(scheme='tableau10'))
)

bars = base.mark_bar()

text = base.mark_text(
    align='center',
    baseline='bottom',
    dy=-5,
    fontWeight='bold',
    color='black',
    fontSize=16
).encode(
    text=altair.Text('Score:Q', format='.3f')
)

combined_chart = (bars + text).properties(
    width=200,
    height=250
).facet(
    column=altair.Column('Domain:N', title=None, header=altair.Header(labelFontWeight='bold', labelFontSize=14)),
    row=altair.Row('Metric:N', title=None, header=altair.Header(labelFontWeight='bold', labelFontSize=14))
)

combined_chart
```

```python
combined_chart.save('figs/rag_perf.png', ppi=500)
```
