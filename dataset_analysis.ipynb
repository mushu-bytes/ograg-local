```python
import tiktoken
import pandas as pd
import altair as alt
from utils import get_documents
```

```python
CLBP_FILE_PATHS = "backpain_data_upgraded_ontology/papers/"
SCI_FILE_PATHS = "/home/damon/scire-pdfs/pdfs"
enc = tiktoken.encoding_for_model("gpt-3.5-turbo")
```

```python
clbp_documents = get_documents(CLBP_FILE_PATHS)
sci_documents = get_documents(SCI_FILE_PATHS)
```

# Number of Documents Per Dataset

```python
len(clbp_documents)
```

```python
len(sci_documents)
```

# Token Distribution

```python
clbp_tokens = pd.DataFrame({"token_count" : list(map(lambda x: len(enc.encode(x.text)) , clbp_documents))})
```

```python
sci_tokens = pd.DataFrame({"token_count" : list(map(lambda x: len(enc.encode(x.text)) , sci_documents))})
```

# Visualization

```python
# Create histogram
clbp_chart = alt.Chart(clbp_tokens).mark_bar().encode(
    alt.X('token_count:Q', bin=True, title='Number of Tokens'),
    alt.Y('count()', title='Number of Documents')
).configure_axis(
      grid=False  # Remove gridlines
).properties(
    width=600,
    height=400,
    title='CLBP: Token Distribution'
)

clbp_chart
```

```python
# Create histogram
sci_chart = alt.Chart(sci_tokens).mark_bar(color='#FF8C42').encode(
    alt.X('token_count:Q', bin=True, title='Number of Tokens'),
    alt.Y('count()', title='Number of Documents')
).configure_axis(
      grid=False  # Remove gridlines
).properties(
    width=600,
    height=400,
    title='SCI: Token Distribution'
)

sci_chart
```

# Save Charts

```python
clbp_chart.save("evaluation/figs/clbp_dataset_token_distro.png", ppi=500)
```
